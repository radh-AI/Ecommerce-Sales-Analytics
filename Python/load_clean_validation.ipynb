```python
reconciliation = sales[
    ["order_id", "gross_sales", "quantity"]
].merge(
    items_summary,
    on="order_id",
    how="left",
    validate="one_to_one"
)

display(reconciliation.head())
```

```python
reconciliation["sales_difference"] = (
    reconciliation["gross_sales"]
    - reconciliation["items_gross_sales"]
)

reconciliation["quantity_difference"] = (
    reconciliation["quantity"]
    - reconciliation["items_quantity"]
)

print(
    "Gross sales mismatches:",
    (reconciliation["sales_difference"].abs() > 0.01).sum()
)

print(
    "Quantity mismatches:",
    (reconciliation["quantity_difference"] != 0).sum()
)
```

```python
order_items_work = order_items.merge(
    sales[["order_id", "quantity"]].rename(
        columns={"quantity": "quantity_sales"}
    ),
    on="order_id",
    how="left",
    validate="many_to_one"
)

order_items_work["cumulative_quantity"] = (
    order_items_work.groupby("order_id")["quantity"].cumsum()
)

order_items_clean = order_items_work[
    order_items_work["cumulative_quantity"]
    <= order_items_work["quantity_sales"]
].copy()

extra_order_items = order_items_work[
    order_items_work["cumulative_quantity"]
    > order_items_work["quantity_sales"]
].copy()

print("Original rows:", len(order_items))
print("Reconciled rows:", len(order_items_clean))
print("Additional rows:", len(extra_order_items))

extra_order_items = order_items_work[
    order_items_work["cumulative_quantity"]
    > order_items_work["quantity_sales"]
].copy()

print("Original rows:", len(order_items))
print("Reconciled rows:", len(order_items_clean))
print("Additional rows:", len(extra_order_items))
```

```python
financial_columns = [
    "quantity",
    "gross_sales",
    "discount_amount",
    "tax_amount",
    "shipping_cost",
    "net_sales",
    "product_cost",
    "profit"
]

clean_summary = (
    order_items_clean
    .groupby("order_id")[financial_columns]
    .sum()
)

validation = sales[
    ["order_id"] + financial_columns
].merge(
    clean_summary,
    on="order_id",
    how="left",
    suffixes=("_sales", "_items"),
    validate="one_to_one"
)

for col in financial_columns:
    difference = (
        validation[f"{col}_sales"]
        - validation[f"{col}_items"]
    )

    if col == "quantity":
        mismatches = difference.ne(0).sum()
    else:
        mismatches = difference.abs().gt(0.01).sum()

    print(f"{col}: {mismatches} mismatches")
```

```python

from pathlib import Path
import pandas as pd

folder = Path("cleaned_data")

files = {
    "customers": "customers_clean.csv",
    "products": "products_clean.csv",
    "sales": "sales_clean.csv",
    "order_items": "order_items_clean.csv"
}

sql_statements = []

for table_name, filename in files.items():
    df = pd.read_csv(folder / filename, nrows=100)

    columns = []

    for col in df.columns:
        safe_col = col.replace('"', '""')
        columns.append(f'    "{safe_col}" TEXT')

    statement = (
        f'CREATE TABLE staging.{table_name} (\n'
        + ",\n".join(columns)
        + "\n);"
    )

    sql_statements.append(statement)

sql_script = (
    "CREATE SCHEMA IF NOT EXISTS staging;\n\n"
    + "\n\n".join(sql_statements)
)

Path("create_staging_tables.sql").write_text(
    sql_script,
    encoding="utf-8"
)

print(sql_script)
print("\nSQL script created successfully!")
```

```python

import os
import re

sql_folder = r"C:\Users\DELL\Downloads\E-comm postgresql"

keywords = {
    "CREATE SCHEMA": r"\bCREATE\s+SCHEMA\b",
    "CREATE TABLE": r"\bCREATE\s+TABLE\b",
    "INSERT INTO": r"\bINSERT\s+INTO\b",
    "UPDATE": r"\bUPDATE\b",
    "ALTER TABLE": r"\bALTER\s+TABLE\b",
    "DELETE FROM": r"\bDELETE\s+FROM\b",
    "SELECT": r"\bSELECT\b",
    "CREATE VIEW": r"\bCREATE\s+(?:OR\s+REPLACE\s+)?VIEW\b"
}

print("SQL SCRIPT CLASSIFICATION")
print("-------------------------")

for filename in sorted(os.listdir(sql_folder)):

    if filename.lower().endswith(".sql"):

        file_path = os.path.join(
            sql_folder,
            filename
        )

        with open(
            file_path,
            "r",
            encoding="utf-8-sig",
            errors="replace"
        ) as file:
            content = file.read()

        # Remove comments before checking SQL operations
        content = re.sub(
            r"/\*.*?\*/",
            "",
            content,
            flags=re.DOTALL
        )

        content = re.sub(
            r"--[^\n]*",
            "",
            content
        )

        detected = []

        for operation, pattern in keywords.items():

            if re.search(
                pattern,
                content,
                flags=re.IGNORECASE
            ):
                detected.append(operation)

        print("\nFILE:", filename)

        print(
            "OPERATIONS:",
            ", ".join(detected)
            if detected else "None detected"
        )
```

```python

from pathlib import Path
import re

file_path = Path(
    r"C:\Users\DELL\Downloads"
    r"\Ecommerce-Sales-Analytics"
    r"\SQL\load_order_items.sql"
)

sql = file_path.read_text(encoding="utf-8-sig")

match = re.search(
    r"INSERT\s+INTO\s+analytics\.order_items\s*"
    r"\((.*?)\)\s*SELECT\s+(.*?)"
    r"\s+FROM\s+staging\.order_items\s*;",
    sql,
    flags=re.IGNORECASE | re.DOTALL
)

print("SQL structure found:", match is not None)

if match:
    insert_columns = [
        col.strip()
        for col in match.group(1).split(",")
    ]

    # Split SELECT expressions only at top-level commas.
    select_text = match.group(2)
    select_expressions = []
    current = []
    depth = 0

    for char in select_text:
        if char == "(":
            depth += 1
        elif char == ")":
            depth -= 1

        if char == "," and depth == 0:
            select_expressions.append(
                "".join(current).strip()
            )
            current = []
        else:
            current.append(char)

    if current:
        select_expressions.append(
            "".join(current).strip()
        )

    duplicates = sorted({
        col for col in insert_columns
        if insert_columns.count(col) > 1
    })

    print("INSERT columns:", len(insert_columns))
    print("SELECT expressions:", len(select_expressions))
    print("Unique INSERT columns:", len(set(insert_columns)))
    print("Duplicate INSERT columns:", duplicates)
    print(
        "Column counts match:",
        len(insert_columns) == len(select_expressions)
    )
    print(
        "Identity column excluded:",
        "order_item_id" not in insert_columns
    )
```

```python

from pathlib import Path
import re
from collections import Counter

file_path = Path(
    r"C:\Users\DELL\Downloads"
    r"\Ecommerce-Sales-Analytics"
    r"\SQL\analytics_schema.sql"
)

sql = file_path.read_text(encoding="utf-8-sig")

# Remove SQL single-line comments.
sql_clean = re.sub(r"--[^\n]*", "", sql)

# Extract named constraints.
constraints = re.findall(
    r"\bADD\s+CONSTRAINT\s+(\w+)\s+"
    r"(PRIMARY\s+KEY|FOREIGN\s+KEY)",
    sql_clean,
    flags=re.IGNORECASE
)

constraint_names = [
    name.lower()
    for name, constraint_type in constraints
]

primary_keys = [
    name
    for name, constraint_type in constraints
    if constraint_type.upper() == "PRIMARY KEY"
]

foreign_keys = [
    name
    for name, constraint_type in constraints
    if constraint_type.upper() == "FOREIGN KEY"
]

counts = Counter(constraint_names)

duplicates = [
    name
    for name, count in counts.items()
    if count > 1
]

print("Total constraints:", len(constraints))
print("Primary keys:", len(primary_keys))
print("Foreign keys:", len(foreign_keys))
print("Duplicate constraint names:", duplicates)

print("\nPrimary-key constraints:")
for name in primary_keys:
    print("-", name)

print("\nForeign-key constraints:")
for name in foreign_keys:
    print("-", name)

print("\nAll checks passed:",
      len(constraints) == 7
      and len(primary_keys) == 4
      and len(foreign_keys) == 3
      and len(duplicates) == 0)
```

```python

from pathlib import Path
import re

sql_folder = Path(
    r"C:\Users\DELL\Downloads"
    r"\Ecommerce-Sales-Analytics\SQL"
)

view_files = [
    "inr_analytical_views.sql",
    "customer segment yearly view.sql",
    "customer yearly view.sql"
]

all_views = {}

# Step 1: Extract each CREATE VIEW statement.
for filename in view_files:

    sql = (sql_folder / filename).read_text(
        encoding="utf-8-sig"
    )

    sql = re.sub(r"--[^\n]*", "", sql)

    matches = re.findall(
        r"\bCREATE\s+(?:OR\s+REPLACE\s+)?VIEW\s+"
        r"analytics\.(\w+)\s+AS\b(.*?);",
        sql,
        flags=re.IGNORECASE | re.DOTALL
    )

    for view_name, definition in matches:
        all_views[view_name.lower()] = {
            "file": filename,
            "definition": definition
        }

# Step 2: Identify dependencies between views.
print("Total views found:", len(all_views))

for view_name, details in all_views.items():

    references = re.findall(
        r"\b(?:FROM|JOIN)\s+analytics\.(\w+)",
        details["definition"],
        flags=re.IGNORECASE
    )

    dependencies = sorted({
        ref.lower()
        for ref in references
        if ref.lower() in all_views
    })

    print(f"\nView: {view_name}")
    print("File:", details["file"])
    print("Depends on views:", dependencies)
```

```python

import pandas as pd
from pathlib import Path

expected_shapes = {
    "customers_clean.csv": (25000, 11),
    "products_clean.csv": (1175, 9),
    "sales_clean.csv": (138116, 46),
    "order_items_clean.csv": (365978, 12)
}

print("EXPORTED CLEANED CSV VALIDATION")
print("-" * 50)

for filename, expected_shape in expected_shapes.items():

    file_path = Path(output_folder) / filename

    df = pd.read_csv(file_path)

    actual_shape = df.shape

    status = (
        "PASS"
        if actual_shape == expected_shape
        else "MISMATCH"
    )

    print(
        f"{filename}: "
        f"{actual_shape[0]:,} rows, "
        f"{actual_shape[1]} columns "
        f"-> {status}"
    )
```

```python

from pathlib import Path

print("CLEANED CSV FILE SIZE CHECK")
print("-" * 50)

total_size = 0

for file_path in sorted(Path(output_folder).glob("*.csv")):

    size_mb = file_path.stat().st_size / (1024 * 1024)

    total_size += size_mb

    print(f"{file_path.name}: {size_mb:.2f} MB")

print("-" * 50)
print(f"Total cleaned CSV size: {total_size:.2f} MB")
```
